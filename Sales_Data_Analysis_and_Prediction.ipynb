## Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

## Create Sales Dataset

```python
data = pd.DataFrame({
    'Month': [1,2,3,4,5,6,7],
    'Sales': [2000,2500,3000,2800,3500,3700,4000]
})

print(data)
```

## Split Features and Target Variable

```python
X = data[['Month']]
y = data['Sales']
```

## Train Linear Regression Model

```python
model = LinearRegression()
model.fit(X, y)
```

## Predict Future Sales

```python
predicted_sales = model.predict([[8]])

print("Predicted Sales for Next Month:", int(predicted_sales[0]))
```

## Perform Numerical Analysis Using NumPy

```python
print("Total Sales:", np.sum(y))
print("Average Sales:", np.mean(y))
```

## Visualize Sales Trend and Prediction

```python
y_pred = model.predict(X)

plt.figure()

# Actual sales data
plt.scatter(data['Month'], data['Sales'], label="Actual Sales")

# Regression line
plt.plot(data['Month'], y_pred, label="Regression Line")

# New prediction point
plt.scatter(8, predicted_sales[0])

# Show X and Y value
plt.text(8, predicted_sales[0],
         f'(8, {int(predicted_sales[0])})')

plt.title("Sales Trend and Prediction")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()

plt.show()
```
