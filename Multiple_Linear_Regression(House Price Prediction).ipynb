# Multiple Linear Regression - House Price Prediction

## Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

## Create Dataset

```python
data = pd.DataFrame({
    'Area': [800,1000,1200,1500,1800],
    'Bedrooms': [2,2,3,3,4],
    'LocationScore': [5,6,7,8,9],
    'Price': [3000,4000,5000,6500,8000]
})

print(data)
```

## Split Features and Target Variable


```python
X = data[['Area','Bedrooms','LocationScore']]
y = data['Price']
```

## Train Multiple Linear Regression Model


```python
model = LinearRegression()
model.fit(X, y)
```

## Predict House Price for New Input

```python
new_price = model.predict([[1300,3,7]])

print("Predicted Price:", new_price[0])
```

## Visualize Actual vs Predicted Prices


```python
y_pred = model.predict(X)

plt.figure()

plt.scatter(y, y_pred)

# Perfect prediction line
plt.plot([min(y), max(y)], [min(y), max(y)])

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted House Prices")

plt.show()
```
