## Import Required Libraries


```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
```

## Create User Movie Dataset

```python
data = pd.DataFrame({
    'Inception':[5,4,0,0],
    'Titanic':[4,3,0,2],
    'Avatar':[0,0,5,4],
    'Interstellar':[5,4,0,0]
}, index=['User1','User2','User3','User4'])

print(data)
```

## Calculate Cosine Similarity

```python
similarity = cosine_similarity(data)

sim_df = pd.DataFrame(
    similarity,
    index=data.index,
    columns=data.index
)

print(sim_df.round(2))
```

## Find Similar Users

```python
similar_users = sim_df.loc['User1'].sort_values(ascending=False)[1:]

print(similar_users.round(2))
```

## Recommend Movies

```python
user1_movies = data.loc['User1']

recommended_movies = data.columns[user1_movies == 0]

print("Recommended Movies:")
print(list(recommended_movies))
```
